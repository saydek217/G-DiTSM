```python
import torch
from ops.models import TSN
from options import parser
from datas import dataset_config

# === Load args and config ===
args = parser.parse_args([])
n_class, args.train_list, args.val_list, args.test_list, prefix = dataset_config.dataset()

args.arch = 'mobilenetv3_deptheca_mega'
args.num_segments = 8
args.shift = False
args.shift_div = 8
args.dropout = 0.5
args.no_partialbn = False
args.shift = True

# === Load checkpoint ===
ckpt_path = 'best.pth.tar'
def load_clean_checkpoint(ckpt_path, model):
    checkpoint = torch.load(ckpt_path, map_location='cpu')
    state_dict = checkpoint['state_dict']
    
    # Clean the keys (remove "module." prefix)
    new_state_dict = {}
    for k, v in state_dict.items():
        new_k = k.replace("module.", "") if k.startswith("module.") else k
        new_state_dict[new_k] = v

    # Load with strict=False to avoid errors due to slight arch changes
    missing, unexpected = model.load_state_dict(new_state_dict, strict=False)

    print(f"✅ Loaded checkpoint from {ckpt_path}")
    print(f"→ Missing keys: {len(missing)}")
    for k in missing: print("  ⤷", k)
    print(f"→ Unexpected keys: {len(unexpected)}")
    for k in unexpected: print("  ⤷", k)

    return model
# === Initialize model ===
model = TSN(n_class, args.num_segments,
            base_model=args.arch,
            dropout=args.dropout,
            partial_bn=not args.no_partialbn,
            is_shift=args.shift, shift_div=args.shift_div)

model_keys = list(model.state_dict().keys())
# Load checkpoint
ckpt = torch.load(ckpt_path, map_location='cpu')
ckpt_keys_raw = list(ckpt['state_dict'].keys())
ckpt_keys = [k.replace("module.", "") if k.startswith("module.") else k for k in ckpt_keys_raw]

# Sort both for better comparison
model_keys.sort()
ckpt_keys.sort()

print(f"\n🔍 Model has {len(model_keys)} keys")
print(f"🔍 Checkpoint has {len(ckpt_keys)} keys (after cleaning)")

# Compare
diff_model_not_in_ckpt = [k for k in model_keys if k not in ckpt_keys]
diff_ckpt_not_in_model = [k for k in ckpt_keys if k not in model_keys]

print(f"\n❗ Keys in model but not in checkpoint: {len(diff_model_not_in_ckpt)}")
for k in diff_model_not_in_ckpt: print("  ⤷", k)

print(f"\n❗ Keys in checkpoint but not in model: {len(diff_ckpt_not_in_model)}")
for k in diff_ckpt_not_in_model: print("  ⤷", k)

```

```python

import time, sys
import torch, cv2, torch.nn.functional as F
import torchvision.transforms as T
from PIL import Image

from ops.models import TSN
from archs.mobilenet_v3_deptheca_mega import MobileNetV3_DepthECAmega

# Config
ckpt_path = 'best.pth.tar'
label_path = 'datas/jester/category.txt'
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
num_segments = 8
use_fp16 = False
frame_queue = []

# Load labels
with open(label_path, encoding='utf-8') as f:
    labels = [l.strip() for l in f if l.strip()]

# Load model
model = TSN(len(labels), num_segments=num_segments,
            base_model='mobilenetv3_deptheca_mega',
            dropout=0.0, partial_bn=True, is_shift=True)
ckpt = torch.load(ckpt_path, map_location=device)
state_dict = ckpt.get('state_dict', ckpt)

clean_dict = {}
for k, v in state_dict.items():
    new_key = k.replace("module.", "") if k.startswith("module.") else k
    clean_dict[new_key] = v

missing, unexpected = model.load_state_dict(clean_dict, strict=False)
print(f"✅ Model loaded: {len(missing)} missing, {len(unexpected)} unexpected")
model.to(device).eval()
if use_fp16: model.half()

# Define transforms
transform = T.Compose([
    T.Resize(256), T.CenterCrop(224),
    T.ToTensor(), T.Normalize([0.485, 0.456, 0.406],
                              [0.229, 0.224, 0.225])
])

# Open webcam
cap = cv2.VideoCapture(0)
cap.set(cv2.CAP_PROP_FRAME_WIDTH, 1280)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 720)
print("🎥 Running... Press Q in the frame to stop")

ema = None
while True:
    ret, frame = cap.read()
    if not ret:
        print("❌ Failed to grab frame")
        break

    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    tensor = transform(Image.fromarray(rgb)).unsqueeze(0).to(device)
    if use_fp16:
        tensor = tensor.half()
    frame_queue.append(tensor)
    if len(frame_queue) > num_segments:
        frame_queue.pop(0)
    if len(frame_queue) < num_segments:
        continue

    input_clip = torch.cat(frame_queue, dim=0).unsqueeze(0)
    input_clip = input_clip.view(1, num_segments, 3, 224, 224)
    output = model(input_clip.to(device))
    if use_fp16:
        output = output.float()

    probs = F.softmax(output, dim=1)[0]
    ema = probs if ema is None else 0.8 * ema + 0.2 * probs
    conf, idx = ema.max(0)
    label = labels[idx]

    cv2.putText(frame, f"{label}", (10, 50), cv2.FONT_HERSHEY_SIMPLEX, 1.2, (0, 255, 0), 2)
    cv2.imshow("🖐️ Gesture Recognition", frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import time
import cv2
import torch
import torch.nn.functional as F
import torchvision.transforms as T
from PIL import Image
from ops.models import TSN
from archs.mobilenet_v3_deptheca_mega import MobileNetV3_DepthECAmega

# === Setup ===
ckpt_path = 'best.pth.tar'
label_path = 'datas/jester/category.txt'
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
num_segments = 8
use_fp16 = False  # Disable to avoid display issues
frame_queue = []

# === Load labels ===
labels = [l.strip() for l in open(label_path, encoding='utf-8') if l.strip()]

# === Load model ===
model = TSN(len(labels), num_segments=num_segments,
            base_model='mobilenetv3_deptheca_mega',
            dropout=0.0, partial_bn=True, is_shift=True)

ckpt = torch.load(ckpt_path, map_location=device)
state_dict = ckpt.get('state_dict', ckpt)
clean_dict = {k.replace("module.", ""): v for k, v in state_dict.items()}
model.load_state_dict(clean_dict, strict=False)

model.to(device).eval()

# === Image transform ===
transform = T.Compose([
    T.Resize(256), T.CenterCrop(224),
    T.ToTensor(),
    T.Normalize([0.485, 0.456, 0.406],
                [0.229, 0.224, 0.225])
])

# === Open camera ===
cap = cv2.VideoCapture(0)
cap.set(cv2.CAP_PROP_FRAME_WIDTH, 1280)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 720)

if not cap.isOpened():
    raise RuntimeError("❌ Cannot open webcam")

cv2.namedWindow("🖐️ Gesture Recognition", cv2.WINDOW_NORMAL)
cv2.resizeWindow("🖐️ Gesture Recognition", 1280, 720)

print("🎥 Webcam feed started. Press 'q' to quit.")

ema = None
prev = time.time()

# === Inference loop ===
while True:
    ret, frame = cap.read()
    if not ret:
        print("⚠️ Failed to grab frame.")
        continue

    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    img = Image.fromarray(rgb)
    tensor = transform(img).unsqueeze(0).to(device)
    frame_queue.append(tensor)

    if len(frame_queue) > num_segments:
        frame_queue.pop(0)
    if len(frame_queue) < num_segments:
        cv2.imshow("🖐️ Gesture Recognition", frame)
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break
        continue

    clip = torch.cat(frame_queue, dim=0).unsqueeze(0)  # (1, T, C, H, W)
    clip = clip.view(1, num_segments, 3, 224, 224)
    with torch.no_grad():
        out = model(clip)
        probs = F.softmax(out, dim=1)[0]
        ema = probs if ema is None else 0.8 * ema + 0.2 * probs
        conf, idx = ema.max(0)
        label = labels[idx]

    now = time.time()
    fps = 1.0 / (now - prev)
    prev = now

    cv2.putText(frame, f"{label}", (30, 60), cv2.FONT_HERSHEY_SIMPLEX, 1.5, (0, 255, 0), 3)
    cv2.putText(frame, f"{conf:.2f}", (30, 110), cv2.FONT_HERSHEY_SIMPLEX, 1.0, (255, 255, 0), 2)
    cv2.putText(frame, f"{fps:.1f} FPS", (30, 160), cv2.FONT_HERSHEY_SIMPLEX, 1.0, (0, 255, 255), 2)

    cv2.imshow("🖐️ Gesture Recognition", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
