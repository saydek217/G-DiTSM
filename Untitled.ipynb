```python
import cv2

# Try MSMF or DSHOW
cap = cv2.VideoCapture(0, cv2.CAP_DSHOW)  # or cv2.CAP_MSMF

if not cap.isOpened():
    print("Cannot open camera")
    exit()

while True:
    ret, frame = cap.read()
    if not ret:
        print("Can't receive frame")
        break
    cv2.imshow('Camera', frame)
    if cv2.waitKey(1) == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
